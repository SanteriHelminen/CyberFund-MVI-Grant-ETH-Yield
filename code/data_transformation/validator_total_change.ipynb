```python
import numpy as np
import pandas as pd
from scipy import stats
```

```python
active_validators_size = pd.read_csv('../int/active_validators_size.csv')
active_validators_category = pd.read_csv('../int/active_validators_category.csv')
active_validators_pool = pd.read_csv('../int/active_validators_pool.csv')
```

```python
active_validators_size_set = active_validators_size
active_validators_size_set['slot'] = active_validators_size_set['slot'] // 300 * 300

# Create a dictionary for aggregation
agg_dict = {col: 'last' for col in active_validators_size_set.columns}

# Group by 'slot' and apply the aggregation
active_validators_size_set = active_validators_size_set.groupby('slot').agg(agg_dict).reset_index(drop=True)

active_validators_size_set
```

```python
# Calculate percent change
active_validators_size_change = active_validators_size_set.pct_change() * 100

# Replace the original DataFrame with the percent change values
active_validators_size_change

# If you want to keep the slot column intact and only apply the percent change to other columns
active_validators_size_change['slot'] = active_validators_size_set['slot']

active_validators_size_change = active_validators_size_change.drop(index=0)

# Display the DataFrame with percent changes
active_validators_size_change
```
