```python
import pandas as pd
import numpy as np
import pandasql as ps
```

```python
validator_exits_size = pd.read_csv('../int/validator_exits_size.csv')
validator_exits_category = pd.read_csv('../int/validator_exits_category.csv')
validator_exits_pool = pd.read_csv('../int/validator_exits_pool.csv')
active_validators_size = pd.read_csv('../int/active_validators_size.csv')
active_validators_category = pd.read_csv('../int/active_validators_category.csv')
active_validators_pool = pd.read_csv('../int/active_validators_pool.csv')
```

```python
validator_exits_size_set = validator_exits_size
validator_exits_size_set['slot'] = ((validator_exits_size_set['slot'] - 1200) // 300) * 300 + 1200

# Create a dictionary for aggregation
agg_dict = {col: 'sum' for col in validator_exits_size_set.columns if col != 'slot'}
agg_dict['slot'] = 'last'

# Group by 'slot' and apply the aggregation
validator_exits_size_set = validator_exits_size_set.groupby('slot').agg(agg_dict).reset_index(drop=True)

validator_exits_size_set
```

```python
validator_exits_category_set = validator_exits_category
validator_exits_category_set['slot'] = ((validator_exits_category_set['slot'] - 1200) // 300) * 300 + 1200

# Create a dictionary for aggregation
agg_dict = {col: 'sum' for col in validator_exits_category_set.columns if col != 'slot'}
agg_dict['slot'] = 'last'

# Group by 'slot' and apply the aggregation
validator_exits_category_set = validator_exits_category_set.groupby('slot').agg(agg_dict).reset_index(drop=True)

validator_exits_category_set
```

```python
validator_exits_pool_set = validator_exits_pool
validator_exits_pool_set['slot'] = ((validator_exits_pool_set['slot'] - 1200) // 300) * 300 + 1200

# Create a dictionary for aggregation
agg_dict = {col: 'sum' for col in validator_exits_pool_set.columns if col != 'slot'}
agg_dict['slot'] = 'last'

# Group by 'slot' and apply the aggregation
validator_exits_pool_set = validator_exits_pool_set.groupby('slot').agg(agg_dict).reset_index(drop=True)

validator_exits_pool_set
```

```python
active_validators_size = active_validators_size[active_validators_size['slot'].isin(validator_exits_size_set['slot'])]
active_validators_size.reset_index(inplace=True)
active_validators_size = active_validators_size.drop(columns=['index'])

active_validators_category = active_validators_category[active_validators_category['slot'].isin(validator_exits_category_set['slot'])]
active_validators_category.reset_index(inplace=True)
active_validators_category = active_validators_category.drop(columns=['index'])

active_validators_pool = active_validators_pool[active_validators_pool['slot'].isin(validator_exits_pool_set['slot'])]
active_validators_pool.reset_index(inplace=True)
active_validators_pool = active_validators_pool.drop(columns=['index'])
```

```python
slots = validator_exits_size_set['slot']
columns = validator_exits_size_set.columns.drop('slot')
validator_exits_size_percentage = validator_exits_size_set[columns].divide(active_validators_size[columns]) * 100
validator_exits_size_percentage['slot'] = slots

validator_exits_size_percentage
```

```python
slots = validator_exits_category_set['slot']
columns = validator_exits_category_set.columns.drop('slot')
validator_exits_category_percentage = validator_exits_category_set[columns].divide(active_validators_category[columns]) * 100
validator_exits_category_percentage['slot'] = slots

validator_exits_category_percentage
```

```python
slots = validator_exits_pool_set['slot']
columns = validator_exits_pool_set.columns.drop('slot')
validator_exits_pool_percentage = validator_exits_pool_set[columns].divide(active_validators_pool[columns]) * 100
validator_exits_pool_percentage['slot'] = slots

validator_exits_pool_percentage
```

```python
validator_exits_size_percentage.to_csv('../int/validator_exits_size_percentage.csv')
validator_exits_category_percentage.to_csv('../int/validator_exits_category_percentage.csv')
validator_exits_pool_percentage.to_csv('../int/validator_exits_pool_percentage.csv')
```
