```python
import pandas as pd
import numpy as np
```

```python
# Read the CSV files
rewards1 = pd.read_csv('../data/rewards_950_999.csv')
rewards2 = pd.read_csv('../data/rewards_1000_1049.csv')
rewards3 = pd.read_csv('../data/rewards_1050_1099.csv')
rewards4 = pd.read_csv('../data/rewards_1100-1139.csv')
rewards5 = pd.read_csv('../data/rewards_1140_1179.csv')
rewards6 = pd.read_csv('../data/rewards_1180_1219.csv')
rewards7 = pd.read_csv('../data/rewards_1220_1259.csv')

# Concatenate the dataframes
rewards = pd.concat([rewards1, rewards2, rewards3, rewards4, rewards5, rewards6, rewards7], ignore_index=True)
rewards = rewards[rewards['earnings'] <= 1000000000]
rewards.reset_index()

rewards
```

```python
del rewards1
del rewards2
del rewards3
del rewards4
del rewards5
del rewards6
del rewards7
```

```python
#1061 238725 225
rewards.loc[rewards['day'] == 1061, 'epoch'] = 238594
```

```python
# Calculate the epoch value for each day based on the known day 1061
base_day = 1061
base_epoch = 238594
epochs_per_day = 225

# Create a function to calculate the epoch for a given day
def calculate_epoch(day):
    day_difference = day - base_day
    return base_epoch + day_difference * epochs_per_day

# Apply the function to calculate the epoch for each day
rewards['epoch'] = rewards['day'].apply(calculate_epoch)

# Display the dataframe to verify the changes
rewards
```

```python
rewards.drop(columns=('day'))
```

```python
rewards['median'] = rewards.groupby('epoch')['earnings'].transform('median')
rewards
```

```python
# Select the rows where 'earnings' are greater than 10000000
priority_n_block_rewards = rewards[rewards['earnings'] > 10000000].copy()
priority_n_block_rewards['priority_n_block_rewards'] = priority_n_block_rewards['earnings'] - priority_n_block_rewards['median']
priority_n_block_rewards = priority_n_block_rewards['priority_n_block_rewards']
```

```python
rewards['priority_n_block_rewards'] = priority_n_block_rewards
rewards = rewards.fillna(0)
rewards
```

```python
validator_metadata = pd.read_csv('../int/validator_metadata_v2.csv')
largest_pools = ('Lido','Coinbase','Binance','Rocketpool','Kraken','OKX','Bitcoin Suisse','Ledger Live','Ether.Fi','Mantle')
validator_metadata.loc[~validator_metadata['pool'].isin(largest_pools), 'pool'] = 'Other Stakers'
validator_metadata
```

```python
rewards = pd.merge(rewards, validator_metadata[['validator_index', 'pool', 'category', 'pool_size_label']], on='validator_index', how='left')
rewards.drop(columns=['day', 'median'])
```
